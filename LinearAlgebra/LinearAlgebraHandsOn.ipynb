# Eigen Value & Eigen Vectors :
fundamental concepts in linear algebra used to understand how complex transformations—represented by matrices—affect space.Eigenvectors are specific directions in space that do not change direction when a mathematical transformation is applied to them. They only get stretched, compressed, or flipped.Eigenvalues are the scalar values that represent exactly how much those eigenvectors are stretched or compressed.
